# Step 4 (exploratory): proactive conflict detection from predicted trajectories

A GRU ensemble (5 members) predicts 3 s ahead from 2 s of history, trained either on clean tracks
(`clean`) or on noisy histories with clean targets (`noise_aware`). For every reference conflict and
an equal number of hard negatives, both agents are predicted from histories ending 3 s before the
critical moment; predicted minimum footprint gap ≤ 1 m is a predicted conflict. `unc_auc_wrong` is
the AUC of the ensemble spread for flagging wrong decisions (0.5 = uninformative).

Run `07_train_predictors.py` then `08_predictor_experiments.py` from the Prompt; this notebook loads results.
This experiment is reported in Supplementary Note S1 of the article. `unc_auc_wrong` is the AUC of the
ensemble spread for flagging wrong conflict decisions (0.5 = chance).

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
import pandas as pd
from IPython.display import Image, display
from sensors_noise.predictor_pipeline import collect_predictor_results

OUT_DIR = Path.cwd().parent / "results" / "04_predictor"
res = collect_predictor_results(OUT_DIR, verbose=False)
p = res["pooled"]
```
